## Cómo usar una celda Markdown

Una celda **Markdown** sirve para escribir **texto con formato** en lugar de código. No se ejecuta como Python: solo se *renderiza* (se muestra con formato).

**Pasos:**

1. Selecciona una celda y presiona <kbd>Esc</kbd> para entrar en **modo comando** (la celda se marca con un borde azul).
2. Presiona <kbd>M</kbd> para convertirla en Markdown (o <kbd>Y</kbd> para volver a celda de código).
   También puedes cambiarla desde el menú desplegable de la barra de herramientas: **Code → Markdown**.
3. Presiona <kbd>Enter</kbd> (o haz doble clic) para **editarla**.
4. Escribe el texto usando la sintaxis Markdown.
5. Presiona <kbd>Shift</kbd> + <kbd>Enter</kbd> para **renderizarla**.

> En cada sección de este cuaderno verás primero **cómo se escribe** (dentro de un recuadro gris) y después **cómo se ve**.

## Encabezados

Se escriben con el símbolo `#` seguido de un **espacio**. Cuantos más `#`, más pequeño es el título (hasta 6 niveles).

**Así se escribe:**

```markdown
# Título nivel 1
## Título nivel 2
### Título nivel 3
#### Título nivel 4
##### Título nivel 5
###### Título nivel 6
```

**Así se ve:**

# Título nivel 1
## Título nivel 2
### Título nivel 3
#### Título nivel 4
##### Título nivel 5
###### Título nivel 6

> ⚠️ **Ojo:** el espacio después de `#` es obligatorio. `#Título` **no** funciona; `# Título` **sí**.

**Sintaxis alternativa** (solo para niveles 1 y 2): subrayar el texto con `=` o `-`.

```markdown
Título nivel 1
==============

Título nivel 2
--------------
```

> 💡 Los encabezados también generan automáticamente el **índice** (*Table of Contents*) del panel lateral de JupyterLab y VS Code.

## Párrafos y saltos de línea

- Un **párrafo nuevo** se crea dejando **una línea en blanco** entre dos bloques de texto.
- Un simple <kbd>Enter</kbd> **no** crea un salto de línea: el texto se une a la línea anterior.
- Para forzar un **salto de línea** dentro del mismo párrafo tienes tres opciones:
  terminar la línea con **dos espacios**, terminarla con una **barra invertida** `\`, o usar la etiqueta HTML `<br>`.

**Así se escribe** (los `··` representan dos espacios, que son invisibles):

```markdown
Esta es la primera línea
y esta se une a la anterior porque solo hay un Enter.

Este es otro párrafo, porque hubo una línea en blanco.

Línea que termina con dos espacios··
siguiente línea.

Línea que termina con barra invertida\
siguiente línea.

Línea con etiqueta HTML<br>siguiente línea.
```

**Así se ve:**

Esta es la primera línea
y esta se une a la anterior porque solo hay un Enter.

Este es otro párrafo, porque hubo una línea en blanco.

Línea que termina con dos espacios  
siguiente línea.

Línea que termina con barra invertida\
siguiente línea.

Línea con etiqueta HTML<br>siguiente línea.

## Formato de texto

| Efecto | Así se escribe | Así se ve |
|:---|:---|:---|
| Cursiva | `*texto*` o `_texto_` | *texto* |
| Negrita | `**texto**` o `__texto__` | **texto** |
| Negrita y cursiva | `***texto***` | ***texto*** |
| Tachado | `~~texto~~` | ~~texto~~ |
| Subrayado (HTML) | `<u>texto</u>` | <u>texto</u> |
| Resaltado (HTML) | `<mark>texto</mark>` | <mark>texto</mark> |
| Subíndice (HTML) | `H<sub>2</sub>O` | H<sub>2</sub>O |
| Superíndice (HTML) | `x<sup>2</sup>` | x<sup>2</sup> |
| Texto pequeño (HTML) | `<small>texto</small>` | <small>texto</small> |
| Tecla (HTML) | `<kbd>Ctrl</kbd> + <kbd>C</kbd>` | <kbd>Ctrl</kbd> + <kbd>C</kbd> |
| Código en línea | `` `print()` `` | `print()` |

Los estilos se pueden **combinar**:

```markdown
Este texto tiene **negrita con *cursiva* dentro**, algo ~~tachado~~ y `código`.
```

Este texto tiene **negrita con *cursiva* dentro**, algo ~~tachado~~ y `código`.

## Listas

### Lista con viñetas (sin orden)

Empieza la línea con `-`, `*` o `+` seguido de un espacio. Para **anidar** elementos, añade **4 espacios** (o <kbd>Tab</kbd>) al inicio.

**Así se escribe:**

```markdown
- Frutas
    - Manzana
    - Pera
        - Pera de agua
- Verduras
* También funciona con asterisco
+ O con el signo más
```

**Así se ve:**

- Frutas
    - Manzana
    - Pera
        - Pera de agua
- Verduras
* También funciona con asterisco
+ O con el signo más

### Lista numerada

Escribe un número seguido de un punto y un espacio.

> 💡 **Truco:** el número que escribas no importa (salvo el primero): Markdown numera automáticamente. Puedes escribir `1.` en todas las líneas y así no tendrás que renumerar si insertas un paso.

**Así se escribe:**

```markdown
1. Abrir Jupyter
1. Crear un cuaderno nuevo
1. Escribir en una celda Markdown
    1. Elegir el tipo de celda
    1. Escribir el texto
1. Renderizar con Shift + Enter
```

**Así se ve:**

1. Abrir Jupyter
1. Crear un cuaderno nuevo
1. Escribir en una celda Markdown
    1. Elegir el tipo de celda
    1. Escribir el texto
1. Renderizar con Shift + Enter

### Listas mixtas y con párrafos

Puedes mezclar listas numeradas y con viñetas, y añadir un párrafo dentro de un elemento (línea en blanco + 4 espacios de sangría).

**Así se escribe:**

```markdown
1. Preparar los datos
    - Cargar el archivo
    - Limpiar valores vacíos
2. Analizar

    Este párrafo pertenece al paso 2.

3. Presentar resultados
```

**Así se ve:**

## Citas

Se crean con `>` al inicio de la línea. Se pueden **anidar** con `>>` y pueden contener cualquier otro formato.

**Así se escribe:**

```markdown
> Esto es una cita.
> Puede ocupar varias líneas.
>
> > Esto es una cita anidada.
>
> — **Autor**, *Nombre del libro*
```

**Así se ve:**

> Esto es una cita.
> Puede ocupar varias líneas.
>
> > Esto es una cita anidada.
>
> — **Autor**, *Nombre del libro*

Las citas también son muy útiles para escribir **notas o avisos**:

```markdown
> 💡 **Nota:** recuerda guardar el cuaderno con Ctrl + S.
```

> 💡 **Nota:** recuerda guardar el cuaderno con Ctrl + S.


Usa la función `print()` para mostrar resultados y `len()` para contar elementos.

### Bloque de código

Usa tres comillas invertidas `` ``` `` antes y después del código. Si escribes el nombre del lenguaje justo después de las primeras comillas, se **colorea la sintaxis** (`python`, `sql`, `bash`, `json`, `html`, `r`, `javascript`…).

**Así se escribe:**

````markdown
```python
def saludar(nombre):
    return f"Hola, {nombre}!"

print(saludar("Mundo"))
```
````

**Así se ve:**

```python
def saludar(nombre):
    return f"Hola, {nombre}!"

print(saludar("Mundo"))
```

Otro ejemplo con SQL:

```sql
SELECT nombre, edad
FROM clientes
WHERE edad > 18
ORDER BY nombre;
```

> 📌 El código dentro de una celda Markdown **solo se muestra, no se ejecuta**. Para ejecutarlo usa una celda de tipo *Code*.

## Imágenes

La sintaxis es igual que la de un enlace, pero con un signo de exclamación `!` delante:

```markdown
![texto alternativo](ruta_o_URL_de_la_imagen "título opcional")
```

- **Texto alternativo:** se muestra si la imagen no carga (y lo leen los lectores de pantalla).
- **Ruta o URL:** puede ser una dirección de internet o un archivo de tu computadora.
- **Título:** opcional; aparece al pasar el ratón por encima.

### Imagen desde una URL (internet)

```markdown
![Logo de Python](https://www.python.org/static/community_logos/python-logo-master-v3-TM.png "Logo de Python")
```

### Imagen desde un archivo de tu computadora

La ruta es **relativa a la carpeta donde está el cuaderno**:

```markdown
![Mi gráfico](grafico.png)                ← misma carpeta que el cuaderno
![Mi gráfico](imagenes/grafico.png)       ← dentro de la subcarpeta "imagenes"
![Mi gráfico](../imagenes/grafico.png)    ← en una carpeta superior
```

> ⚠️ Usa siempre `/` en las rutas (también en Windows).
> Si el nombre del archivo tiene espacios, cámbialos por `%20` (`mi%20foto.png`) o usa HTML: `<img src="mi foto.png">`.
>
> 💡 Los **GIF animados** (`.gif`) y las imágenes `.jpg`, `.png` y `.svg` se insertan exactamente igual.

## Tablas

Las columnas se separan con barras verticales `|` y el encabezado se separa del contenido con una fila de guiones `-`. Los dos puntos `:` en esa fila indican la **alineación** de cada columna:

| En la fila de guiones | Alineación |
|:---|:---|
| `:---` | Izquierda |
| `:---:` | Centrada |
| `---:` | Derecha |

**Así se escribe:**

```markdown
| Producto  | Cantidad | Precio |
|:----------|:--------:|-------:|
| Manzanas  |    10    |   2.50 |
| Peras     |    5     |   3.75 |
| **Total** |  **15**  | **6.25** |
```

**Así se ve:**

| Producto  | Cantidad | Precio |
|:----------|:--------:|-------:|
| Manzanas  |    10    |   2.50 |
| Peras     |    5     |   3.75 |
| **Total** |  **15**  | **6.25** |

> 💡 No hace falta que las columnas queden alineadas en el texto: `|A|B|` + `|-|-|` funciona igual. Para escribir una barra `|` dentro de una celda usa `\|`.

## Fórmulas matemáticas con LaTeX

Jupyter interpreta fórmulas escritas en **LaTeX**.

### Fórmula en línea (dentro del texto)

Se escribe entre signos de dólar simples `$ ... $`:

```markdown
La ecuación de Einstein es $E = mc^2$ y el área de un círculo es $A = \pi r^2$.
```

La ecuación de Einstein es $E = mc^2$ y el área de un círculo es $A = \pi r^2$.

### Fórmula en bloque (centrada, en su propia línea)

Se escribe entre dobles signos de dólar `$$ ... $$`:

```markdown
$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$
```

$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$

```markdown
        &= a^2 + 2ab + b^2
\end{aligned}
$$
```

## HTML dentro de Markdown

Las celdas Markdown aceptan muchas etiquetas **HTML**, lo que permite hacer cosas que Markdown solo no puede (colores, tamaños, centrar, desplegables…).

> ⚠️ Por seguridad, Jupyter "limpia" parte del HTML: etiquetas como `<script>` o `<iframe>`, y algunos estilos, pueden no funcionar según el entorno (JupyterLab, VS Code, Colab, GitHub…).

### Color y tamaño de texto

```html
<span style="color:red">Texto rojo</span>,
<span style="color:#1E90FF">azul con código hexadecimal</span>,
<span style="background-color:yellow; color:black">fondo amarillo</span> y
<span style="font-size:22px">texto grande</span>.
```

<span style="color:red">Texto rojo</span>,
<span style="color:#1E90FF">azul con código hexadecimal</span>,
<span style="background-color:yellow; color:black">fondo amarillo</span> y
<span style="font-size:22px">texto grande</span>.

### Texto centrado

```html
<p align="center"><b>Este texto está centrado</b></p>
```

<p align="center"><b>Este texto está centrado</b></p>

### Sección desplegable

Ideal para ocultar soluciones de ejercicios o detalles largos. Deja una **línea en blanco** después de `</summary>` para poder usar Markdown dentro.

```html
<details>
<summary>👉 Haz clic para ver la respuesta</summary>

¡Esta es la respuesta oculta! Aquí dentro puedes usar **Markdown**, listas, imágenes, etc.

</details>
```

<details>
<summary>👉 Haz clic para ver la respuesta</summary>

¡Esta es la respuesta oculta! Aquí dentro puedes usar **Markdown**, listas, imágenes, etc.

</details>

### Comentarios ocultos

Todo lo que esté entre `<!--` y `-->` **no se muestra** al renderizar la celda. Útil para dejarte notas a ti mismo.

```html
<!-- Este texto no se verá -->
```

Debajo de esta línea hay un comentario que no ves (haz doble clic en la celda para encontrarlo).

<!-- ¡Me encontraste! Este es un comentario oculto. -->

### Videos

Markdown no puede incrustar un video de YouTube, pero puedes mostrar su **miniatura como enlace** (cambia `ID_DEL_VIDEO` por el código que aparece en la URL después de `watch?v=`):

```markdown
[![Ver el video](https://img.youtube.com/vi/ID_DEL_VIDEO/0.jpg)](https://www.youtube.com/watch?v=ID_DEL_VIDEO)
```

Para un video guardado en tu computadora:

```html
<video src="mi_video.mp4" controls width="400"></video>
```

## Caracteres especiales y emojis

### Escapar símbolos

Si quieres mostrar literalmente un símbolo que Markdown usa para dar formato, ponle una **barra invertida** `\` delante:

| Así se escribe | Así se ve |
|:---|:---|
| `\*no es cursiva\*` | \*no es cursiva\* |
| `\_no es cursiva\_` | \_no es cursiva\_ |
| `\# no es un título` | \# no es un título |
| `\- no es una lista` | \- no es una lista |
| `\[no es un enlace\]` | \[no es un enlace\] |
| `\$50 dólares` | \$50 dólares |
| `\<b\> no es HTML` | \<b\> no es HTML |

### Entidades HTML (símbolos)

| Así se escribe | Así se ve | | Así se escribe | Así se ve |
|:---|:---:|---|:---|:---:|
| `&copy;` | &copy; | | `&rarr;` | &rarr; |
| `&reg;` | &reg; | | `&larr;` | &larr; |
| `&trade;` | &trade; | | `&uarr;` | &uarr; |
| `&euro;` | &euro; | | `&check;` | &check; |
| `&deg;` | &deg; | | `&hearts;` | &hearts; |
| `&frac12;` | &frac12; | | `&infin;` | &infin; |
| `&lt;` | &lt; | | `&gt;` | &gt; |

> 💡 `&nbsp;` inserta un espacio "duro": sirve para poner varios espacios seguidos, ya que Markdown junta los espacios normales en uno solo.

### Emojis

Copia y pega el emoji directamente en el texto: 😀 🚀 📊 📈 ✅ ❌ ⚠️ 💡 📌 🔥

- En **Windows**: <kbd>Win</kbd> + <kbd>.</kbd> abre el selector de emojis.
- En **macOS**: <kbd>Ctrl</kbd> + <kbd>Cmd</kbd> + <kbd>Espacio</kbd>.

> ⚠️ Los códigos tipo `:smile:` (de GitHub o Slack) **no** funcionan en Jupyter.

## Atajos de teclado

En **modo comando** (después de presionar <kbd>Esc</kbd>):

| Atajo | Acción |
|:---|:---|
| <kbd>M</kbd> | Convertir la celda en **Markdown** |
| <kbd>Y</kbd> | Convertir la celda en **Código** |
| <kbd>R</kbd> | Convertir la celda en **Raw** (texto sin procesar) |
| <kbd>1</kbd> … <kbd>6</kbd> | Convertir en encabezado de nivel 1 a 6 |
| <kbd>Enter</kbd> | Entrar en modo edición |
| <kbd>A</kbd> / <kbd>B</kbd> | Insertar celda arriba / abajo |
| <kbd>D</kbd> <kbd>D</kbd> | Eliminar la celda (presionar D dos veces) |
| <kbd>Z</kbd> | Deshacer la eliminación de una celda |
| <kbd>C</kbd> / <kbd>X</kbd> / <kbd>V</kbd> | Copiar / cortar / pegar celdas |
| <kbd>Shift</kbd> + <kbd>M</kbd> | Unir las celdas seleccionadas |

En cualquier modo:

| Atajo | Acción |
|:---|:---|
| <kbd>Shift</kbd> + <kbd>Enter</kbd> | Renderizar y pasar a la siguiente celda |
| <kbd>Ctrl</kbd> + <kbd>Enter</kbd> | Renderizar y quedarse en la misma celda |
| <kbd>Alt</kbd> + <kbd>Enter</kbd> | Renderizar e insertar una celda nueva debajo |
| <kbd>Ctrl</kbd> + <kbd>Shift</kbd> + <kbd>-</kbd> | Dividir la celda en el cursor (modo edición) |
| <kbd>Ctrl</kbd> + <kbd>S</kbd> | Guardar el cuaderno |
