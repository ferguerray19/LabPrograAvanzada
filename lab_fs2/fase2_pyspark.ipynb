### 0.3 Justificación del `.env`
Regla de oro de la guía: `WORKERS × WORKER_MEMORY + DRIVER_MEMORY` debe quedar holgadamente
bajo la RAM que tiene Docker. Dentro del contenedor, `/proc/meminfo` muestra justamente la RAM
de la máquina virtual de Docker (WSL 2).

```python
def a_gb(txt):
    txt = txt.strip().lower()
    return float(txt[:-1]) / (1024 if txt.endswith("m") else 1)

with open("/proc/meminfo") as f:
    ram_docker_gb = int(f.readline().split()[1]) / 1024**2

# Workers y su memoria se leen del master (lo mismo que muestra la UI :8080), es decir,
# SPARK_WORKERS y SPARK_WORKER_MEMORY tal como quedaron aplicados, no un valor escrito a mano.
estado    = json.load(urlopen("http://spark-master:8080/json/"))
vivos     = [w for w in estado["workers"] if w["state"] == "ALIVE"]
workers   = len(vivos)
w_mem     = max(w["memory"] for w in vivos) / 1024          # MB -> GB (SPARK_WORKER_MEMORY)
w_cores   = max(w["cores"] for w in vivos)
drv_mem   = a_gb(os.environ.get("SPARK_DRIVER_MEMORY", "2g"))
reservado = sum(w["memory"] for w in vivos) / 1024 + drv_mem
exec_mem  = a_gb(os.environ.get("SPARK_EXECUTOR_MEMORY", "2g"))
assert exec_mem <= w_mem, "SPARK_EXECUTOR_MEMORY no cabe en SPARK_WORKER_MEMORY: no habría executors"

print(f"RAM asignada a Docker        : {ram_docker_gb:.1f} GB  ({os.cpu_count()} CPUs)")
print(f"Workers x memoria + driver   : {workers} x {w_mem:g} GB + {drv_mem:g} GB = {reservado:g} GB")
print(f"Margen para el SO y Jupyter  : {ram_docker_gb - reservado:.1f} GB")
print(f"Perfil aplicado              : {workers} workers x {w_cores} cores x {w_mem:g} GB, driver {drv_mem:g} GB")
if ram_docker_gb >= 12:
    # Con 16 GB la guía sugiere 3 workers x 3g; se mantiene el de 8 GB a propósito:
    # todas las mediciones del notebook se hicieron con 4 cores, y el margen libre
    # queda para Docker Desktop, el navegador y la UI de Spark.
    print(f"Nota: Docker tiene {ram_docker_gb:.0f} GB; se usa el perfil de 8 GB de forma conservadora "
          f"(margen de {ram_docker_gb - reservado:.0f} GB) para que las mediciones sean comparables.")
assert reservado < ram_docker_gb, "El clúster pide más memoria de la que tiene Docker"
M["ram_docker_gb"] = round(ram_docker_gb, 1)
```

### 0.4 Cómo trabaja el clúster (implicancia 1)

Una sola máquina tiene un techo: su RAM, sus cores y su disco. Si los datos no caben, en una
máquina la única salida es comprar una más grande (escalado vertical), y eso tiene límite.
Spark escala **horizontalmente**: divide los datos en particiones y cada worker procesa las
suyas en paralelo. Para crecer se agregan workers (`docker compose up -d --scale spark-worker=3`).



* **Driver** (este notebook, contenedor `jupyter`): ejecuta el programa, arma el plan de cada
  consulta, lo divide en tareas y las reparte. No procesa el volumen distribuido (eso lo hacen los
  executors), pero sí trabaja con los resultados chicos que recibe: en este notebook, pandas y
  matplotlib corren en el driver.
* **Master**: sabe qué workers hay y le asigna recursos al driver. Tampoco procesa datos.
* **Workers**: cada uno corre un *executor* con 2 cores y 2 GB. Ahí se ejecutan las tareas.

Prueba de humo con `verificar_cluster.py`. Se corre antes de abrir el notebook, porque esta
sesión ocupa los 4 cores del clúster y otra aplicación quedaría esperando recursos. Las dos
capturas que siguen son del 29-09; antes de la ejecución final del notebook la prueba se volvió
a correr y pasó.



UI del master (:8080) con los 2 workers registrados:



**¿Por qué un `.collect()` de millones de filas es un error?** `collect()` trae todas las filas
al driver, que es una sola máquina con 2 GB. El crudo son 1,75 GB de texto y como objetos de
Python ocuparía varias veces eso: el driver se queda sin memoria, o Spark corta antes por
`spark.driver.maxResultSize` (1 GB por defecto). Además se pierde el clúster, porque el trabajo
vuelve a una sola máquina. Por eso en este notebook al driver solo llegan resultados ya
agregados, de pocas filas (`show()`, `first()`, `toPandas()` de tablas chicas).


### 0.5 Herramientas de medición
Tres funciones pequeñas que se usan en todo el notebook:

* `medir(df)`: ejecuta la consulta completa **sin traer datos al driver** (formato `noop`) y
  devuelve la mediana de varias repeticiones. Nunca se mide con `.collect()`: eso mediría
  también la red hacia el driver.
* `contar_exchanges(df)`: cuenta los `Exchange` del plan físico, separados en dos tipos. Un
  `Exchange` de reparto (`hashpartitioning`, `rangepartitioning`...) es un **shuffle**: todas
  las filas se redistribuyen por la red. Un `BroadcastExchange` **no** es un shuffle: copia una
  tabla chica entera a cada executor y la tabla grande no se mueve.
* `tareas(grupo)`: consulta la API REST de la UI (puerto 4040) y devuelve cuánto tardó cada
  tarea. Es lo mismo que se ve en la pestaña *Stages*, pero en una tabla. Cada ejecución lleva
  una etiqueta propia y cada etapa se cuenta una sola vez, así que repetir una celda no mezcla
  la corrida nueva con la anterior.

```python
def medir(df, reps=3):
    """Mediana (s) de ejecutar la consulta entera sin traer filas al driver."""
    tiempos = []
    for _ in range(reps):
        t0 = time.perf_counter()
        df.write.format("noop").mode("overwrite").save()
        tiempos.append(time.perf_counter() - t0)
    return round(statistics.median(tiempos), 2)


def plan_fisico(df):
    """Plan físico hasta la tabla cacheada: lo que hay debajo de InMemoryRelation es el plan
    de la limpieza (ya ejecutado y guardado en memoria), no el de esta consulta."""
    plan = df._jdf.queryExecution().executedPlan().toString()
    salida, nivel_oculto = [], None
    for linea in plan.splitlines():
        nivel = next((i for i, ch in enumerate(linea) if ch.isalnum()), len(linea))
        if nivel_oculto is not None and nivel > nivel_oculto:
            continue                      # rama interna de la tabla cacheada: se omite
        nivel_oculto = None
        salida.append(linea)
        if "InMemoryRelation" in linea:
            nivel_oculto = nivel
    return "\n".join(salida)


def mostrar_plan(df):
    for linea in plan_fisico(df).splitlines():
        if linea.strip(" :+-"):
            print(linea)


def contar_exchanges(df):
    plan = plan_fisico(df)
    shuffles   = len(re.findall(r"(?<!Broadcast)Exchange (hashpartitioning|rangepartitioning|RoundRobinPartitioning|SinglePartition)", plan))
    broadcasts = len(re.findall(r"BroadcastExchange", plan))
    return shuffles, broadcasts


ultima_etiqueta = {}   # grupo -> etiqueta de su última ejecución


def ejecutar_en_grupo(grupo, df):
    """Ejecuta df etiquetando sus jobs, para poder buscarlos después en la UI. Cada ejecución
    tiene una etiqueta única (grupo#uuid): repetir una celda, o esta misma, no mezcla corridas."""
    ultima_etiqueta[grupo] = f"{grupo}#{uuid.uuid4().hex}"
    sc.setJobGroup(ultima_etiqueta[grupo], grupo)
    try:
        t0 = time.perf_counter()
        df.write.format("noop").mode("overwrite").save()
        dur = time.perf_counter() - t0
    finally:
        sc.setJobGroup("otros", "otros")
    time.sleep(2)            # la UI registra las métricas de forma asíncrona
    return round(dur, 2)


def tareas(grupo):
    """Duración y filas leídas del shuffle de cada tarea de la última ejecución del grupo."""
    base = f"{sc.uiWebUrl}/api/v1/applications/{sc.applicationId}"
    jobs = sc.statusTracker().getJobIdsForGroup(ultima_etiqueta[grupo])
    # Con AQE, varios jobs de la misma consulta comparten etapas: cada etapa se cuenta una vez.
    etapas = sorted({sid for jid in jobs for sid in sc.statusTracker().getJobInfo(jid).stageIds})
    por_tarea = {}
    for sid in etapas:
        try:
            intentos = json.load(urlopen(f"{base}/stages/{sid}?details=false"))
        except HTTPError as e:
            if e.code == 404:
                continue     # etapa saltada: nunca se ejecutó
            raise
        for intento in sorted(intentos, key=lambda a: a["attemptId"]):
            url = f"{base}/stages/{sid}/{intento['attemptId']}/taskList?length=100000"
            for t in json.load(urlopen(url)):
                if t.get("status") == "SUCCESS":
                    por_tarea[(sid, t["index"])] = t   # si se reintentó, queda el último intento
    filas = []
    for (sid, _), t in sorted(por_tarea.items()):
        m = t.get("taskMetrics") or {}
        filas.append(dict(etapa=sid,
                          dur_ms=m.get("executorRunTime", 0),
                          filas_shuffle=(m.get("shuffleReadMetrics") or {}).get("recordsRead", 0)))
    return pd.DataFrame(filas)


def resumen_etapa_shuffle(grupo):
    """De las etapas que LEEN un shuffle, la de mayor volumen: nº de tareas, mediana y máximo."""
    t = tareas(grupo)
    t = t[t.filas_shuffle > 0]
    etapa = t.groupby("etapa").filas_shuffle.sum().idxmax()
    e = t[t.etapa == etapa]
    med = max(e.dur_ms.median(), 1)
    return dict(etapa=int(etapa), tareas_con_shuffle=len(e),
                mediana_ms=int(e.dur_ms.median()), max_ms=int(e.dur_ms.max()),
                ratio_max_mediana=round(e.dur_ms.max() / med, 1),
                max_filas=int(e.filas_shuffle.max()),
                mediana_filas=int(e.filas_shuffle.median())), e

print("Herramientas listas")
```

### 1.1 Esquema inferido vs esquema declarado (implicancia 3)
Si Spark lee JSON **sin esquema**, primero recorre *todo* el archivo para adivinar los tipos
y después lo vuelve a leer para procesarlo: dos pasadas. Con el esquema declarado hay una sola.

```python
RAW = "/opt/workspace/data/raw/*.jsonl"

esquema_raw = StructType([
    StructField("timestamp",  TimestampType(), True),   # ISO 8601 con "Z": Spark lo convierte solo
    StructField("sensor_id",  StringType(), True),
    StructField("metric",     StringType(), True),
    StructField("value",      DoubleType(), True),
    StructField("unit",       StringType(), True),
    StructField("worker_id",  LongType(),   True),
    StructField("site",       StringType(), True),
    StructField("asset_type", StringType(), True),
    StructField("status",     StringType(), True),
    StructField("seq",        LongType(),   True),
])

# (a) Inferido: el propio spark.read.json ya lanza un job que lee TODO para adivinar tipos.
t0 = time.perf_counter()
df_inferido = spark.read.json(RAW)
t_inferencia = time.perf_counter() - t0
t0 = time.perf_counter()
n_inf = df_inferido.count()
t_conteo_inf = time.perf_counter() - t0

# (b) Declarado: leer no cuesta nada (evaluación perezosa); solo el count ejecuta.
t0 = time.perf_counter()
raw = spark.read.schema(esquema_raw).json(RAW)
t_decl_lectura = time.perf_counter() - t0
t0 = time.perf_counter()
n_raw = raw.count()
t_conteo_decl = time.perf_counter() - t0

assert n_inf == n_raw
print(pd.DataFrame([
    ["inferido",  round(t_inferencia, 2),  round(t_conteo_inf, 2),  round(t_inferencia + t_conteo_inf, 2)],
    ["declarado", round(t_decl_lectura, 2), round(t_conteo_decl, 2), round(t_decl_lectura + t_conteo_decl, 2)],
], columns=["esquema", "spark.read (s)", "count (s)", "total (s)"]).to_string(index=False))
print(f"\nRegistros: {n_raw:,}")
print(f"Ahorro por declarar el esquema: {t_inferencia + t_conteo_inf - t_decl_lectura - t_conteo_decl:.1f} s "
      f"({(t_inferencia + t_conteo_inf) / (t_decl_lectura + t_conteo_decl):.1f}x)")
M.update(raw_registros=n_raw, t_inferido=round(t_inferencia + t_conteo_inf, 2),
         t_declarado=round(t_decl_lectura + t_conteo_decl, 2))
```

```python
# ¿Qué tipos adivinó Spark? La inferencia deja timestamp como texto (string); el esquema
# declarado lo tipa como timestamp. Y no hay garantía de estabilidad: un archivo con un
# "value": "NaN" como texto cambiaría la columna entera a string.
print("Inferido :", [(f.name, f.dataType.simpleString()) for f in df_inferido.schema.fields])
print("Declarado:", [(f.name, f.dataType.simpleString()) for f in raw.schema.fields])
```

**Lectura.** Con el esquema inferido, `spark.read` ya tarda varios segundos: Spark recorrió los
10 millones de eventos solo para adivinar los tipos, y después el `count` los vuelve a leer.
Con el esquema declarado, `spark.read` es casi instantáneo y la única lectura es la del `count`.
Además, con el esquema declarado los tipos no dependen de lo que traigan los archivos: si un
día llega un valor mal escrito, el inferido puede cambiar el tipo de toda la columna entre una
ejecución y otra; el declarado no.

Hay una diferencia más en los tipos: la inferencia deja `timestamp` como **texto**, porque JSON no
tiene un tipo fecha. Con el esquema declarado queda como `timestamp` desde la lectura, así que se
puede comparar, filtrar por rango y particionar por fecha sin castear después. Si algún valor no
tuviera el formato esperado, Spark lo dejaría nulo: por eso el perfilado (sección 2) verifica con
una aserción que `timestamp` no tenga nulos.

**En Big Data:** aquí la diferencia son segundos; con 1 TB, inferir es leer 1 TB extra en cada
ejecución.


### 1.2 Evaluación perezosa (implicancia 2)

En Spark hay dos tipos de operaciones:
* **Transformaciones** (`filter`, `withColumn`, `select`, `groupBy`...): solo agregan un paso al
  plan. No leen datos.
* **Acciones** (`count`, `show`, `write`, `collect`...): obligan a ejecutar. Spark optimiza el plan
  y corre lo que esa acción necesita para su resultado.

Ya se vio en 1.1: `spark.read` con esquema declarado tardó casi 0 s porque todavía no leyó nada.
Aquí se arma una consulta con tres transformaciones, se mira su plan y recién al final se ejecuta.


```python
t0 = time.perf_counter()
alarmas = (raw.filter(F.col("status") == "ALARM")
              .withColumn("value_f", F.col("value") * 9 / 5 + 32)
              .select("timestamp", "sensor_id", "value_f"))
t_transf = time.perf_counter() - t0

alarmas.explain()

t0 = time.perf_counter()
n_alarmas = alarmas.count()
t_accion = time.perf_counter() - t0

print(f"\nTransformaciones (filter, withColumn, select): {t_transf:.3f} s")
print(f"Acción (count)                               : {t_accion:.2f} s  ->  {n_alarmas:,} alarmas")
M.update(t_lazy_transf=round(t_transf, 3), t_lazy_accion=round(t_accion, 2))
```

**Lectura.** Las tres transformaciones tardan una fracción de segundo: Spark solo anotó qué hacer. El tiempo
real aparece en el `count`, que es la acción. Eso es lo que demuestra esta celda: la ejecución
perezosa.

El `explain()` muestra el plan de la consulta tal como está escrita, el que correría un `show()` o
un `write`. Ahí se ve por qué conviene esperar. En la línea `FileScan json` aparece `PushedFilters`
con el filtro por `status`, y `ReadSchema` trae solo las 4 columnas que la consulta usa, no las 10.
Como Spark conoce la consulta completa antes de ejecutarla, mueve el filtro al momento de la
lectura y deja de lado las columnas que nunca se usan. En JSON igual se recorre todo el texto,
pero Spark solo convierte esas columnas y descarta las filas apenas las lee; en Parquet (1.4) el
ahorro es mayor porque se salta columnas completas en disco.

Ojo: el `count()` **no** ejecuta exactamente ese plan. Para contar filas no hacen falta
`timestamp`, `sensor_id` ni `value_f`, así que Spark vuelve a optimizar la consulta para el
`count`: lee solo `status`, filtra y cuenta, sin calcular `value_f`. El tiempo del `count` mide
leer y filtrar el JSON, no todo lo que muestra el `explain()` de arriba. Su plan propio se ve con
`alarmas.groupBy().count().explain()`.

**En Big Data:** si cada línea se ejecutara apenas se escribe, habría que guardar resultados
intermedios de millones de filas. Con el plan completo, Spark hace todo en una sola pasada.


### 1.3 ¿Cuántas particiones de lectura y por qué?
Spark corta los archivos en trozos de `spark.sql.files.maxPartitionBytes` (128 MB por
defecto). Cada trozo es una **partición** y cada partición la procesa **una tarea**.

```python
import glob, math
bytes_raw = sum(os.path.getsize(p) for p in glob.glob(RAW))
max_part = 128 * 1024**2
bloques = [math.ceil(os.path.getsize(p) / max_part) for p in sorted(glob.glob(RAW))]
print(f"Tamaño del crudo        : {bytes_raw / 1024**3:.2f} GB en {len(glob.glob(RAW))} archivos")
print(f"maxPartitionBytes       : {spark.conf.get('spark.sql.files.maxPartitionBytes')}")
print(f"Total / 128 MB          : {bytes_raw / max_part:.1f}  (solo referencia)")
print(f"Bloques por archivo     : {bloques}  -> {sum(bloques)} trozos antes de agruparlos")
print(f"Particiones reales      : {raw.rdd.getNumPartitions()}")
print(f"Cores del clúster       : {CORES}  -> {raw.rdd.getNumPartitions() / CORES:.1f} 'olas' de tareas")
M["raw_particiones"] = raw.rdd.getNumPartitions()
```

**Lectura.** ¿Por qué 15 y no 14? Spark no divide el total: corta **cada archivo por separado**
en bloques de 128 MB. Cada archivo de ~450 MB da 3 bloques completos más un resto (≈66 MB en los
tres primeros, ≈56 MB en el último): 16 trozos. Después los **empaqueta** en particiones de hasta
128 MB, y dos de los restos chicos caben juntos en una sola. Por eso quedan 15 particiones, que con
4 cores son casi 4 "olas" de tareas: en la última ola sobran cores esperando.

### 1.4 Capa *bronze*: el mismo crudo, en Parquet
JSON es texto: para leer una sola columna hay que parsear la línea entera. Parquet es
**columnar y comprimido**. Se convierte una vez y todo el perfilado se hace sobre la copia en
Parquet.

```python
BRONZE = "/opt/workspace/salida_fase2/bronze_raw.parquet"
t0 = time.perf_counter()
raw.write.mode("overwrite").parquet(BRONZE)
t_escritura = time.perf_counter() - t0

bronze = spark.read.parquet(BRONZE)
bytes_bronze = sum(os.path.getsize(p) for p in glob.glob(f"{BRONZE}/*.parquet"))

consulta_json    = raw.groupBy("metric").agg(F.avg("value"))
consulta_parquet = bronze.groupBy("metric").agg(F.avg("value"))
t_json, t_parquet = medir(consulta_json, 1), medir(consulta_parquet, 2)   # JSON: 1 pasada, cada una lee 1,8 GB

print(f"Conversión JSON -> Parquet : {t_escritura:.1f} s (se paga una vez)")
print(f"Tamaño JSON / Parquet      : {bytes_raw/1024**2:,.0f} MB / {bytes_bronze/1024**2:,.0f} MB "
      f"({bytes_raw/bytes_bronze:.1f}x más chico)")
print(f"Promedio por métrica       : JSON {t_json} s  vs  Parquet {t_parquet} s "
      f"({t_json/t_parquet:.1f}x más rápido)")
M.update(mb_json=round(bytes_raw/1024**2), mb_parquet=round(bytes_bronze/1024**2),
         t_consulta_json=t_json, t_consulta_parquet=t_parquet)
```

**En Big Data:** la conversión se paga una vez; después cada consulta lee una fracción de los bytes.

## 5. Limpieza con criterio

| Regla | Qué se hace | Por qué |
|---|---|---|
| 1 | Eliminar los **duplicados exactos** | Son la misma lectura registrada dos veces; no aportan información |
| 2 | **No** deduplicar por clave | Borraría ~110 mil muestras válidas de la báscula |
| 3 | **Marcar** las lecturas degradadas (`pegado`, `deriva`, `asentamiento`), no borrarlas | En la Fase 3 el detector debe aprender a distinguir falla de *instrumento* vs falla de *proceso*; si se borran, esa información desaparece |
| 4 | Mantener los nulos estructurales de `alimentador_id` | "No aplica" no es "cero" |
| 5 | Tipos: `alimentador_id` a entero, también en la verdad de terreno | Para poder unir por esa columna |
| 6 | Verificar dominios con aserciones | Si algo cae fuera de rango, que falle aquí y no semanas después |

```python
limpio = (
    tele.dropDuplicates()                                              # regla 1
        .withColumn("alimentador_id", F.col("alimentador_id").cast("int"))  # regla 5
        .withColumn("sensor_confiable", F.col("calidad_sensor") == "ok")    # regla 3
)
verdad_limpia = verdad.withColumn("alimentador_afectado",
                                  F.col("alimentador_afectado").cast("int"))

# Regla 6: dominios físicos. Se CUENTA lo que queda fuera, no se filtra en silencio.
fuera_de_rango = limpio.filter(
    ((F.col("variable") == "temperatura_c") & ~F.col("valor").between(-40, 60)) |
    ((F.col("variable") == "humedad_pct")   & ~F.col("valor").between(0, 100)) |
    ((F.col("variable") == "velocidad_ms")  & (F.col("valor") < 0))
).count()

limpio = limpio.cache()          # se reutiliza muchas veces en la sección 6
n_limpio = limpio.count()

assert n_limpio == n_tele - dup_exactos_tele, "Se perdieron o ganaron filas en la limpieza"
assert fuera_de_rango == 0, f"{fuera_de_rango} lecturas fuera de dominio físico"

# Una comparación con NULL no es verdadera ni falsa: el filtro de rangos no ve los nulos.
# Se cuentan aparte en las columnas obligatorias (alimentador_id no: sus nulos son "no aplica").
obligatorias = ["timestamp_utc", "entidad_id", "variable", "valor", "calidad_sensor"]
nulos_obligatorios = limpio.filter(
    " OR ".join(f"{c} IS NULL" for c in obligatorias)).count()
calidades = {r[0] for r in limpio.select("calidad_sensor").distinct().collect()}
assert nulos_obligatorios == 0, f"{nulos_obligatorios} filas con nulos en {obligatorias}"
assert calidades <= {"ok", "asentamiento", "deriva", "pegado"}, f"calidad_sensor desconocida: {calidades}"

print(f"Filas antes            : {n_tele:,}")
print(f"Duplicados eliminados  : {dup_exactos_tele}")
print(f"Filas después          : {n_limpio:,}")
print(f"Fuera de dominio       : {fuera_de_rango}")
print(f"Nulos en obligatorias  : {nulos_obligatorios}")
limpio.groupBy("sensor_confiable", "calidad_sensor").count().orderBy("calidad_sensor").show()
M.update(limpio_filas=n_limpio,
         no_confiables=limpio.filter(~F.col("sensor_confiable")).count())
```

**Lectura.** La limpieza quitó solo lo que era un error de verdad: los 2 duplicados exactos del
instante inicial (6.395.325 → 6.395.323 filas). Todo lo demás se conservó a propósito:
* Las ~110 mil lecturas repetidas por clave de la báscula siguen ahí, porque son muestras válidas.
* Las 124.219 lecturas degradadas (`asentamiento`, `deriva`, `pegado`) quedaron **marcadas** con
  `sensor_confiable = false`, no borradas, para que la Fase 3 pueda distinguir falla de instrumento
  de falla de proceso.
* Los nulos de `alimentador_id` se mantienen, porque significan "no aplica".

Las aserciones confirman que no quedó nada fuera de dominio físico ni nulos en las columnas
obligatorias. La tabla limpia queda en caché para la sección 6.


## 6. Shuffling: el concepto central de la fase · implicancia 5

**En Big Data:** un shuffle escribe datos a disco, los manda por la red a otro worker y los vuelve
a leer. Es lo más caro de un job, así que gran parte de optimizar Spark es evitar shuffles o
achicarlos.

### 6.1 Operaciones estrechas vs anchas
* **Estrechas** (`filter`, `withColumn`, `select`): cada partición se procesa sola. 0 shuffles.
* **Anchas** (`groupBy`, `join`, `distinct`, `orderBy`): las filas con la misma clave tienen
  que juntarse en el mismo lugar, así que viajan por la red. En el plan, cada `Exchange` de reparto (`hashpartitioning`,
  `rangepartitioning`...) es un shuffle. El `BroadcastExchange` es otra cosa: copia una tabla
  chica a todos los executors sin mover la grande (sección 6.6).

```python
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)   # para ver el join "caro"
consultas = {
    "filter + withColumn": limpio.filter("valor > 0").withColumn("doble", F.col("valor") * 2),
    "groupBy":             limpio.groupBy("entidad_id").agg(F.avg("valor")),
    "distinct":            limpio.select("entidad_id").distinct(),
    "orderBy":             limpio.orderBy("valor"),
    "join (sort-merge)":   limpio.join(limpio.select("entidad_id", "entidad_tipo").distinct(), "entidad_id"),
}
tabla = [(nombre, *contar_exchanges(df)) for nombre, df in consultas.items()]
spark.conf.unset("spark.sql.autoBroadcastJoinThreshold")
print(pd.DataFrame(tabla, columns=["operación", "shuffles (Exchange)", "broadcasts"]).to_string(index=False))
```

### 6.2 El default de 200 particiones de shuffle
Después de un shuffle, Spark reparte los datos en `spark.sql.shuffle.partitions` particiones
(200 por defecto). Con 4 cores eso son 50 "olas" de tareas diminutas: casi todo el tiempo se
va en administrar tareas. Se mide la misma agregación con tres configuraciones.

```python
agregacion = lambda: limpio.groupBy("entidad_id", "variable").agg(
    F.avg("valor").alias("media"), F.min("valor").alias("min"),
    F.max("valor").alias("max"), F.count("*").alias("n"))

n_grupos = agregacion().count()
configs = [("200 particiones, sin AQE", 200, "false"),
           (f"{2*CORES} particiones (2x cores), sin AQE", 2 * CORES, "false"),
           (f"{4*CORES} particiones (4x cores), sin AQE", 4 * CORES, "false"),
           ("200 particiones, con AQE", 200, "true")]
filas = []
for nombre, part, aqe in configs:
    spark.conf.set("spark.sql.shuffle.partitions", part)
    spark.conf.set("spark.sql.adaptive.enabled", aqe)
    medir(agregacion(), 1)                         # calentamiento
    t = medir(agregacion(), 3)
    salida = agregacion().rdd.getNumPartitions()   # con AQE: las que quedaron tras fusionar
    filas.append((nombre, t, salida))
spark.conf.unset("spark.sql.shuffle.partitions")
spark.conf.set("spark.sql.adaptive.enabled", "true")

res_part = pd.DataFrame(filas, columns=["configuración", "tiempo (s)", "particiones de salida"])
print(f"Agregación de {n_limpio:,} filas en {n_grupos} grupos, {CORES} cores\n")
print(res_part.to_string(index=False))
mejor = res_part["tiempo (s)"][1:3].idxmin()      # la más rápida entre 2x y 4x cores
mejora = res_part["tiempo (s)"][0] / res_part["tiempo (s)"][mejor]
print(f"\nAjustar el parámetro ({res_part['configuración'][mejor]}): "
      f"{mejora:.1f}x más rápido que 200 sin tocar la lógica.")
M.update(n_grupos=n_grupos, t_200=filas[0][1], t_8=filas[1][1], t_16=filas[2][1],
         t_aqe=filas[3][1], part_aqe=filas[3][2], mejora_particiones=round(mejora, 1))
```

```python
import matplotlib.pyplot as plt
os.makedirs("/opt/workspace/salida_fase2", exist_ok=True)

fig, ax = plt.subplots(figsize=(7, 3))
ax.barh(res_part["configuración"], res_part["tiempo (s)"], color=["#b0b0b0", "#2a7ab9", "#5b9bd5", "#6aa84f"])
for i, v in enumerate(res_part["tiempo (s)"]):
    ax.text(v, i, f" {v} s", va="center")
ax.invert_yaxis(); ax.set_xlabel("segundos (mediana de 3)")
ax.set_title("Misma agregación, distinto nº de particiones de shuffle")
fig.tight_layout(); fig.savefig("/opt/workspace/salida_fase2/grafico_particiones.png", dpi=120)
plt.show()
```

**Lectura.** La agregación produce 329 grupos. Con 200 particiones casi todas las tareas reciben
muy pocas filas, y Spark gasta más tiempo creando y coordinando tareas que calculando. Con 8
(2 por core) cada core hace 2 tareas con trabajo real, y con 16 (4 por core), 4 tareas. La regla
práctica es quedarse entre 2 y 4 veces los cores (cuál de las dos gana cambia de una corrida a
otra, porque quedan cerca; las dos son varias veces más rápidas que 200): menos deja cores sin trabajo si una tarea se
atrasa, y muchas más vuelve a gastar el tiempo en coordinar tareas. Con AQE, Spark mide el shuffle al
terminar y junta solo las particiones chicas, sin que nadie ajuste el número. AQE mejora el
default sin tocar código, pero el shuffle igual se escribe en 200 trozos y se juntan recién al
leer; por eso un valor bien elegido a mano puede quedar igual o mejor.


### 6.3 AQE (Adaptive Query Execution)
Con AQE activo (viene así en `docker/spark-defaults.conf`), Spark mira el tamaño **real** de
cada partición al terminar el shuffle y **fusiona** las diminutas. Por eso en la tabla anterior
las 200 particiones pedidas terminan convertidas en muy pocas, sin que nadie ajustara nada.
AQE resuelve buena parte del problema solo, pero no todos los casos (ver 6.4).

### 6.4 Sesgo de claves (*data skew*) y el *straggler* · implicancia 6

```python
top = por_entidad.limit(3).toPandas()
mediana_entidad = por_entidad.approxQuantile("count", [0.5], 0)[0]
print(top.to_string(index=False))
print(f"\nMediana de filas por entidad: {mediana_entidad:,.0f}")
print(f"faja_1 / mediana            : {top['count'][0] / mediana_entidad:.0f} a 1")
M.update(faja_pct=float(top["pct"][0]), faja_filas=int(top["count"][0]),
         ratio_faja_mediana=round(top["count"][0] / mediana_entidad))
```

¿Por qué afecta? Si se agrupa por `entidad_id`, **todas** las filas de `faja_1` van a la misma
partición y las procesa **una sola tarea**. La etapa termina cuando termina la tarea más lenta:
los otros cores quedan esperando. Esa tarea lenta es el *straggler*.

Un `groupBy` con promedios casi no lo sufre, gracias a la agregación parcial (cada partición
envía un resumen). Donde el sesgo **sí** duele es en operaciones que necesitan *todas* las
filas de la clave juntas: las **funciones de ventana** y los **joins**.

Ejemplo: la **diferencia** de cada lectura respecto de la anterior (`valor - lag(valor)`), que
necesita todas las filas de la clave ordenadas en una sola tarea. Se usa aquí **para mostrar el
sesgo**, no como la variable final de la Fase 3, por dos razones:
* Es una diferencia, no una tasa: para una tasa habría que dividir por el tiempo transcurrido
  entre las dos lecturas.
* La báscula tiene varias muestras con el mismo `timestamp_utc` (sección 4), así que ordenar
  solo por tiempo no define cuál va primero entre ellas. Para la Fase 3 haría falta una columna
  que desempate (por ejemplo, un número de secuencia).

```python
spark.conf.set("spark.sql.shuffle.partitions", 32)
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "false")

def diferencia_con_anterior(claves):
    w = Window.partitionBy(*claves).orderBy("timestamp_utc")
    return limpio.withColumn("delta", F.col("valor") - F.lag("valor").over(w))

t_a = ejecutar_en_grupo("ventana_entidad", diferencia_con_anterior(["entidad_id"]))
r_a, tareas_a = resumen_etapa_shuffle("ventana_entidad")
t_b = ejecutar_en_grupo("ventana_entidad_variable", diferencia_con_anterior(["entidad_id", "variable"]))
r_b, tareas_b = resumen_etapa_shuffle("ventana_entidad_variable")

res_skew = pd.DataFrame([
    dict(particion_por="entidad_id", tiempo_s=t_a, **r_a),
    dict(particion_por="entidad_id + variable", tiempo_s=t_b, **r_b),
]).drop(columns="etapa")
print(res_skew.to_string(index=False))
M.update(skew_a=r_a, skew_b=r_b, t_ventana_a=t_a, t_ventana_b=t_b)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3), sharey=True)
for ax, (tit, e) in zip(axes, [("partitionBy(entidad_id)", tareas_a),
                               ("partitionBy(entidad_id, variable)", tareas_b)]):
    e = e.sort_values("dur_ms", ascending=False).reset_index(drop=True)
    ax.bar(e.index, e.dur_ms / 1000, color="#2a7ab9")
    ax.set_title(tit); ax.set_xlabel("tareas (ordenadas por duración)")
axes[0].set_ylabel("segundos por tarea")
fig.suptitle("Duración de cada tarea de la etapa de ventana: el straggler es la barra alta")
fig.tight_layout(); fig.savefig("/opt/workspace/salida_fase2/grafico_straggler.png", dpi=120)
plt.show()
```

**Lectura.** Con `partitionBy(entidad_id)` una sola tarea recibe los ~3 millones de filas de
`faja_1` mientras la tarea típica recibe ~100 mil: en nuestra ejecución tardó **unas 14 veces
la mediana**. Esa barra solitaria del gráfico es el *straggler*; en la UI se ve en la pestaña
*Stages* → *Summary Metrics*: la fila *Max* de *Duration* muy por encima de *Median*.

Además, particionar solo por entidad era **incorrecto**: mezclaba variables distintas
(restaba una velocidad de una corriente). La clave correcta `(entidad_id, variable)` reparte
la faja en 5 claves: la tarea más cargada baja de ~3,2 M a ~0,7 M filas y el máximo/mediana
baja de ~14x a ~8x. Es la mitigación más barata: **una clave más fina que siga teniendo sentido**.

Lo estable es el **reparto** (filas por tarea, máximo/mediana), no el tiempo total. Con tareas de
pocos segundos, el ruido de la máquina pesa tanto como el straggler: en las corridas anteriores la
clave fina terminó antes (por ejemplo, 10,4 s → 6,4 s), y en esta tardó algo más (10,9 s → 12,1 s).
Con más datos, el straggler domina el tiempo y la mejora del reparto se nota en el total.

**En la UI (:4040, pestaña Stages).** Etapa de la ventana por `entidad_id`, capturada mientras
corría el job:



En *Summary Metrics*, la mediana de *Shuffle Read Records* es 100.800 filas y el máximo
3.175.200 (faja_1): exactamente las mismas filas que la tabla de arriba. La captura es de una
ejecución anterior, así que sus **tiempos** no son los de la tabla (ahí la mediana de *Duration*
era 0,1 s y el máximo 3 s); los tiempos cambian de una corrida a otra, pero el patrón es el mismo:
una sola tarea tarda decenas de veces más que la mediana. Esa tarea es además la única con
*spill*: sus datos no cupieron en la memoria de ejecución y Spark los bajó a disco. La UI muestra
ese mismo derrame con dos cifras: *Spill (memory)* = 384 MiB es lo que ocupaban esos datos en
memoria (deserializados) y *Spill (disk)* = 39,4 MiB es lo que ocuparon ya escritos en disco
(serializados y comprimidos). No son dos derrames distintos.

**En Big Data:** agregar workers no arregla el sesgo, porque todas las filas de una clave van a
una sola tarea. Hay que cambiar la clave: una más fina (como aquí) o *salting* (6.5).


### 6.5 *Salting* (y cómo AQE lo hace solo en los joins)
Cuando no hay una clave más fina con sentido, se **inventa**: a cada fila de la tabla grande se
le agrega una "sal" aleatoria `0..N-1` y la tabla chica se replica N veces (una copia por sal).
Así `faja_1` se reparte en N tareas. Se prueba con un *sort-merge join* entre la telemetría y
una tabla de activos (se desactiva el broadcast a propósito, para forzar el shuffle).

```python
N_SAL = 8
dim_activos = (limpio.select("entidad_id", "entidad_tipo").distinct()
                     .withColumn("zona", F.when(F.col("entidad_tipo") == "reefer", "patio")
                                          .otherwise("frente_atraque")))

spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)
spark.conf.set("spark.sql.adaptive.autoBroadcastJoinThreshold", -1)
spark.conf.set("spark.sql.adaptive.skewJoin.enabled", "false")

join_normal = limpio.join(dim_activos.drop("entidad_tipo"), "entidad_id")
t_jn = ejecutar_en_grupo("join_normal", join_normal)
r_jn, _ = resumen_etapa_shuffle("join_normal")

grande_sal = limpio.withColumn("sal", (F.rand(seed=7) * N_SAL).cast("int"))
chica_sal  = dim_activos.drop("entidad_tipo").crossJoin(
    spark.range(N_SAL).select(F.col("id").cast("int").alias("sal")))
join_sal = grande_sal.join(chica_sal, ["entidad_id", "sal"]).drop("sal")
t_js = ejecutar_en_grupo("join_sal", join_sal)
r_js, _ = resumen_etapa_shuffle("join_sal")


spark.conf.set("spark.sql.adaptive.skewJoin.enabled", "true")
spark.conf.set("spark.sql.adaptive.skewJoin.skewedPartitionThresholdInBytes", "4MB")
spark.conf.set("spark.sql.adaptive.skewJoin.skewedPartitionFactor", "2")
spark.conf.set("spark.sql.adaptive.advisoryPartitionSizeInBytes", "8MB")   # tamaño de cada trozo
t_ja = ejecutar_en_grupo("join_aqe", limpio.join(dim_activos.drop("entidad_tipo"), "entidad_id"))
r_ja, _ = resumen_etapa_shuffle("join_aqe")

assert join_normal.count() == join_sal.count() == n_limpio

for k in ["spark.sql.autoBroadcastJoinThreshold", "spark.sql.adaptive.autoBroadcastJoinThreshold",
          "spark.sql.adaptive.skewJoin.skewedPartitionThresholdInBytes",
          "spark.sql.adaptive.skewJoin.skewedPartitionFactor",
          "spark.sql.adaptive.advisoryPartitionSizeInBytes", "spark.sql.shuffle.partitions"]:
    spark.conf.unset(k)
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "true")

res_sal = pd.DataFrame([
    dict(variante="sort-merge normal", tiempo_s=t_jn, **r_jn),
    dict(variante=f"sort-merge con sal (N={N_SAL})", tiempo_s=t_js, **r_js),
    dict(variante="sort-merge + AQE skewJoin", tiempo_s=t_ja, **r_ja),
]).drop(columns="etapa")
print(res_sal.to_string(index=False))
M.update(sal_normal=r_jn, sal_sal=r_js, sal_aqe=r_ja, t_join_normal=t_jn, t_join_sal=t_js, t_join_aqe=t_ja)
```

**Lectura.**
* **Con sal**, la tarea más cargada pasa de ~3,2 M filas a ~0,5 M (columna `max_filas`) y la
  relación máximo/mediana cae de ~15x a ~4x. El tiempo total también baja, pero es la cifra
  menos estable entre corridas (aquí 18,3 s → 9,5 s; en otra corrida, 8,6 s → 7,9 s), porque
  lo que se ahorra en el straggler se gasta en parte en replicar la tabla chica N veces.
  **El salting tiene un costo y solo compensa cuando el sesgo es grande**, y siempre hay que
  comprobar que el resultado no cambió (la aserción).
* **AQE `skewJoin`** hace lo mismo automáticamente: detecta la partición gigante y la parte en
  trozos (por eso sube el número de tareas). Sin tocar la consulta dejó la
  tarea más cargada más chica (~0,4 M filas) y obtuvo el menor tiempo. Viene activado en `spark-defaults.conf`; aquí solo se bajaron sus umbrales
  porque, por defecto, considera "gigante" una partición de más de 256 MB.

### 6.6 *Broadcast join* · implicancia 7
Si una de las tablas es pequeña, lo inteligente es **no** hacer shuffle: se copia la tabla
chica entera a cada executor y la tabla grande se queda donde está.

```python
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)
spark.conf.set("spark.sql.adaptive.autoBroadcastJoinThreshold", -1)
smj = limpio.join(dim_activos, "entidad_id")
ex_smj = contar_exchanges(smj); t_smj = medir(smj)
spark.conf.unset("spark.sql.autoBroadcastJoinThreshold")
spark.conf.unset("spark.sql.adaptive.autoBroadcastJoinThreshold")

bhj = limpio.join(F.broadcast(dim_activos), "entidad_id")
ex_bhj = contar_exchanges(bhj); t_bhj = medir(bhj)

print(pd.DataFrame([["sort-merge join", *ex_smj, t_smj], ["broadcast join", *ex_bhj, t_bhj]],
                   columns=["tipo", "shuffles", "broadcasts", "tiempo (s)"]).to_string(index=False))
print(f"\nBroadcast: {t_smj / t_bhj:.1f}x más rápido; la tabla grande no se movió.")
M.update(t_smj=t_smj, t_bhj=t_bhj, ex_smj=ex_smj[0], ex_bhj=ex_bhj[0])
```

**Lectura.** El sort-merge necesita 3 shuffles: el del `distinct` de la tabla chica y uno por cada
tabla para juntar las filas con el mismo `entidad_id`. Con broadcast, la tabla chica (una fila
por entidad) se copia entera a cada executor y la tabla grande no se mueve: queda solo el shuffle
del `distinct` y el join es más rápido.

**Cuándo usar cada uno:** broadcast cuando una tabla cabe holgada en la memoria de cada executor
(Spark lo hace solo bajo 10 MB, `spark.sql.autoBroadcastJoinThreshold`); sort-merge cuando las
dos tablas son grandes.


### 6.7 `repartition` vs `coalesce`

```python
rep = limpio.repartition(2)
coa = limpio.coalesce(2)
print(pd.DataFrame([
    ["repartition(2)", contar_exchanges(rep)[0], medir(rep), rep.rdd.getNumPartitions()],
    ["coalesce(2)",    contar_exchanges(coa)[0], medir(coa), coa.rdd.getNumPartitions()],
], columns=["operación", "shuffles", "tiempo (s)", "particiones"]).to_string(index=False))
```

Las dos terminan con 2 particiones, pero `repartition` paga un shuffle completo (redistribuye
todas las filas) y `coalesce` solo junta particiones vecinas sin mover datos por la red.
**Para bajar particiones: `coalesce`. Para reequilibrar datos sesgados o subir: `repartition`.**

### 6.8 La trampa del *fan-out* en los joins
Objetivo: etiquetar cada lectura con la falla que estaba ocurriendo (si había alguna), para
que la Fase 3 pueda medir precisión y *recall*.

**Versión ingenua:** unir por entidad.

```python
ingenuo = limpio.join(F.broadcast(verdad_limpia),
                      limpio["entidad_id"] == verdad_limpia["entidad_afectada"], "left")
n_ingenuo = ingenuo.count()
print(f"Filas antes del join : {n_limpio:,}")
print(f"Filas después        : {n_ingenuo:,}   <- {n_ingenuo - n_limpio:,} filas fabricadas de la nada")

try:
    assert n_ingenuo == n_limpio, "El join alteró el número de filas"
except AssertionError as e:
    print(f"\nAssertionError atrapado: {e}")

etiq_faja = (ingenuo.filter("entidad_id = 'faja_1' AND falla_id IS NOT NULL")
                    .dropDuplicates(limpio.columns).count())      # lecturas distintas marcadas
total_faja = limpio.filter("entidad_id = 'faja_1'").count()
print(f"Lecturas de faja_1 marcadas como falla: {etiq_faja:,} de {total_faja:,} "
      f"({etiq_faja / total_faja:.0%}) -> ¡incluye los días en que funcionó perfecto!")
M.update(filas_ingenuo=n_ingenuo)
```

Dos errores a la vez:
1. **Filas duplicadas.** `faja_1` tuvo dos fallas (`rodamiento_faja` y `sobrecarga_faja`), así
   que cada lectura de la faja se copió dos veces.
2. **Etiquetas falsas.** Una falla dura horas, no la semana entera; unir solo por entidad
   marca como "en falla" el 100 % de sus lecturas.

A esto se suma algo propio de este dataset: las fallas de **alimentador** (`alimentador_2`,
`alimentador_3`) también afectan a los 25 reefers conectados a ese alimentador, que el join
ingenuo no etiqueta.

**Versión correcta: *range join*.** Coincide la entidad (o el alimentador) **y** el instante cae
dentro de la ventana de la falla. La ventana es semiabierta, `t_inicio <= t < t_fin`, igual que en
el simulador (`generar_salida_sitio3.py`): en `t_fin` la falla ya terminó.

```python
v = verdad_limpia.select("falla_id", "tipo_falla", "entidad_afectada", "alimentador_afectado",
                         "t_inicio", "t_fin", "severidad")
t = limpio
cond = (
    ((t["entidad_id"] == v["entidad_afectada"]) |
     (v["alimentador_afectado"].isNotNull() & (t["alimentador_id"] == v["alimentador_afectado"])))
    & (t["timestamp_utc"] >= v["t_inicio"])
    & (t["timestamp_utc"] < v["t_fin"])          # t_fin excluido, como en el simulador
)
etiquetada = (t.join(F.broadcast(v), cond, "left")
               .drop("entidad_afectada", "alimentador_afectado", "t_inicio", "t_fin")
               .withColumn("en_falla", F.col("falla_id").isNotNull()))

n_etiquetada = etiquetada.count()
assert n_etiquetada == n_limpio, "El join alteró el número de filas"
print(f"Filas antes: {n_limpio:,} | después: {n_etiquetada:,}  -> la aserción pasa")

print("\nLecturas etiquetadas por falla:")
etiquetada.filter("en_falla").groupBy("falla_id", "tipo_falla").agg(
    F.count("*").alias("lecturas"), F.countDistinct("entidad_id").alias("entidades")
).orderBy("falla_id").show(truncate=False)

etiq_faja_ok = etiquetada.filter("entidad_id = 'faja_1' AND en_falla").count()
print(f"faja_1 en falla: {etiq_faja_ok:,} de {total_faja:,} lecturas ({etiq_faja_ok / total_faja:.0%})")
mostrar_plan(etiquetada)
M.update(filas_etiquetada=n_etiquetada, lecturas_en_falla=etiquetada.filter("en_falla").count(),
         pct_faja_ingenuo=round(etiq_faja / total_faja * 100), pct_faja_correcto=round(etiq_faja_ok / total_faja * 100))
```

El plan muestra `BroadcastNestedLoopJoin`: como la condición tiene un `OR` y rangos, no hay
una clave de igualdad para un *hash join*; con solo 9 filas en la tabla chica, comparar cada
lectura contra las 9 fallas en memoria es barato y **no hay shuffle de la tabla grande**.

**¿Y la falla F005?** La verdad de terreno tiene 9 fallas, pero la tabla muestra 8. F005
(`transferencia_generador` en `alimentador_3`) duró solo **21 segundos** (21:09:32 → 21:09:53 del
17-01), y los reefers y alimentadores se registran cada 60 s: ninguna lectura cayó dentro de esa
ventana. El join no la perdió; simplemente no hay lecturas que etiquetar. Para la Fase 3 es un
aviso: una falla más corta que el periodo de muestreo puede no dejar ninguna lectura dentro de su
ventana, como F005. Un detector solo podría notarla por sus efectos en lecturas posteriores, si
los hay.

## 7. Escritura de la capa limpia y verificación · implicancia 8

Se escribe en Parquet **particionado por `fecha`**: 7 carpetas, una por día, lo que habilita
el *partition pruning* de la sección 3.1 en la Fase 3.

¿Por qué no particionar por `timestamp_utc`? Porque hay cientos de miles de instantes distintos:
serían cientos de miles de carpetas con archivos diminutos. Leerlas sería lentísimo (cada
archivo tiene un costo fijo de apertura) y el sistema de archivos sufriría.

Para que cada día quede en un solo archivo, y no en uno por tarea, se usa `coalesce(1)` antes de
escribir. `coalesce` junta particiones que ya existen sin hacer shuffle (sección 6.7: 0 `Exchange`).
`repartition` también lo lograría, pero movería las 6,4 M filas por la red solo para escribirlas.
El costo de `coalesce(1)` es que la escritura la hace una sola tarea; con este volumen (unos 50 MB
en Parquet) no importa, y con mucho más se usaría `coalesce(n)` con un n mayor.


```python
SALIDA = "/opt/workspace/salida_fase2/telemetria_limpia.parquet"
n_instantes = limpio.select("timestamp_utc").distinct().count()
print(f"Instantes distintos: {n_instantes:,}  (serían {n_instantes:,} carpetas si se particionara por timestamp)")

t0 = time.perf_counter()
(etiquetada
    .coalesce(1)
    .write.mode("overwrite")
    .partitionBy("fecha")
    .parquet(SALIDA))
print(f"Escritura: {time.perf_counter() - t0:.1f} s")
```

```python
releida = spark.read.parquet(SALIDA)

# Verificaciones con aserciones: si algo falla, el notebook se detiene aquí.
assert releida.count() == n_limpio, "El total escrito no coincide"

por_dia_antes   = {r.fecha: r["count"] for r in etiquetada.groupBy("fecha").count().collect()}
por_dia_despues = {r.fecha: r["count"] for r in releida.groupBy("fecha").count().collect()}
assert por_dia_antes == por_dia_despues, "Algún día perdió o ganó filas"

assert releida.filter("en_falla").count() == M["lecturas_en_falla"], "Se perdieron etiquetas"
assert releida.filter("NOT sensor_confiable").count() == M["no_confiables"], "Se perdieron marcas de calidad"
assert set(releida.columns) >= {"timestamp_utc", "entidad_id", "variable", "valor",
                                "sensor_confiable", "falla_id", "en_falla"}

tipos_escritos = {c.name: c.dataType for c in etiquetada.schema if c.name != "fecha"}
tipos_releidos = {c.name: c.dataType for c in releida.schema if c.name != "fecha"}
assert tipos_escritos == tipos_releidos, "El esquema cambió al releer"
# fecha no va dentro del archivo: Spark la reconstruye desde el nombre de la carpeta.
assert isinstance(releida.schema["fecha"].dataType, DateType), "fecha no se releyó como date"

print("Todas las aserciones pasaron.\n")
print(f"{'carpeta':<22}{'archivos':>9}{'MB':>8}{'filas':>12}")
for d in sorted(os.listdir(SALIDA)):
    ruta = os.path.join(SALIDA, d)
    if os.path.isdir(ruta):
        archivos = [f for f in os.listdir(ruta) if f.endswith(".parquet")]
        mb = sum(os.path.getsize(os.path.join(ruta, f)) for f in archivos) / 1024**2
        fecha = d.split("=")[1]
        print(f"{d:<22}{len(archivos):>9}{mb:>8.1f}{por_dia_despues[pd.Timestamp(fecha).date()]:>12,}")
releida.printSchema()
```

**Lectura.** La relectura pasa todas las aserciones: mismas filas en total y por día, mismas
etiquetas, mismas marcas de calidad y mismos tipos de columna. Quedan 7 carpetas `fecha=...` con
un archivo cada una. Particionado por `timestamp_utc` serían 604.800 carpetas con archivos
diminutos, y abrir cada archivo tiene un costo fijo que haría la lectura lentísima.
